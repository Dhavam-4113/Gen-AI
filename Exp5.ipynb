```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-1.5B-Instruct",
    torch_dtype="auto",
    device_map="auto"
)

def generate_response(prompt, max_tokens=200):
    messages = [
        {
            "role": "system",
            "content": "You are a helpful AI assistant that follows instructions accurately."
        },
        {
            "role": "user",
            "content": prompt
        }
    ]

    result = generator(
        messages,
        max_new_tokens=max_tokens,
        do_sample=False
    )

    return result[0]["generated_text"][-1]["content"]
```

```python
print("="*60)
print("PROMPT ENGINEERING APPLICATION")
print("="*60)

########################################################
# CONTENT GENERATION
########################################################

content_prompt = """
Write exactly five simple sentences introducing Generative Artificial Intelligence.

Requirements:
- Use simple English.
- Audience: First-year engineering students.
- Mention chatbots.
- Mention image generation.
- Avoid repetition.
"""

print("\n1. CONTENT GENERATION")
print("-"*60)
print(generate_response(content_prompt))

########################################################
# REASONING
########################################################

reasoning_prompt = """
A college conducted a Generative AI workshop for 120 students.

85 students completed the workshop.

Calculate:

1. Total students
2. Completed students
3. Students who did not complete

Show the subtraction.
"""

print("\n2. REASONING TASK")
print("-"*60)
print(generate_response(reasoning_prompt))

########################################################
# EMAIL
########################################################

email_prompt = """
Write a formal email.

Subject: Generative AI Laboratory Session

Include:

Friday

10:00 AM

AI Laboratory 2

Students must bring laptops.

Students must complete Hugging Face registration before attending.

Keep it professional.
"""

print("\n3. EMAIL GENERATION")
print("-"*60)
print(generate_response(email_prompt))

########################################################
# ACTION ITEMS
########################################################

meeting_notes = """
The project team reviewed the development of the college chatbot.

Arun will prepare the training dataset by Monday.

Priya will test the chatbot responses by Wednesday.

Rahul will prepare the final demonstration and presentation.

The team will meet again on Friday to review the progress.
"""

action_prompt = f"""
Extract all action items.

Meeting Notes:

{meeting_notes}

Output exactly like this:

1. Person - Task - Deadline
"""

print("\n4. ACTION ITEM EXTRACTION")
print("-"*60)
print(generate_response(action_prompt))
```
